```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt('stb_plot.dat')

# Extract x and p columns
x = data[:, 0]  # First column (x values)
p = data[:, 1]  # Second column (p values)

plt.figure(figsize=(8, 6))

# Plot the data with dots
plt.plot(x, p, 'ko', markersize=0.2)  # 'bo' means blue dots

# Add labels and title
plt.xlabel('X values')
plt.ylabel('P values')
plt.title('Plot of X vs P')

# Show the grid
plt.grid(True)

# Show the plot
plt.show()
```
